```python
import pandas as pd
import plotly.express as px
```

```python
df=pd.read_csv("C:/Users/Admin/Downloads/student_performance_plotly.csv")
df
```

```python
print(df.head())
```

```python
# Bar chart
fig = px.bar(df,x="Student",y="Marks",
             title="Student Marks",color="Subject")
fig.show()
```

```python
# Line chart
fig = px.line(df, x="Student", y="Marks",
              markers=True,title="Marks Line Chart")
fig.show()
```

```python
# scatter plot
fig = px.scatter(df, x="Study_Hours", y="Marks",
                 color="Subject",
                 title="Study Hours vs Marks")
fig.show()
```

```python
# Pie chart
subject_count = df["Subject"].value_counts().reset_index()
subject_count.columns = ["Subject", "Students"]

fig = px.pie(subject_count, names="Subject",
             values="Students",
             title="Students by Subject")
fig.show()
```

```python
# Histogram
fig = px.histogram(df, x="Marks",
                   title="Marks Distribution",color="Subject")
fig.show()
```

```python
# Box plot
fig = px.box(df, x="Subject", y="Marks",
             title="Marks by Subject",color="Subject")
fig.show()
```

```python
# Violine plot

fig = px.violin(df, x="Subject", y="Marks",
                box=True, points="all",
                title="Marks Spread by Subject", color="Subject")
fig.update_layout(
    width=800,
    height=500
)
fig.show()
```

```python
# Area chart
fig = px.area(df, x="Student", y="Marks",
              title="Student Marks Area Chart", color="Subject")
fig.show()
```

```python
# Average marks bar chart
avg_marks = df.groupby("Subject", as_index=False)["Marks"].mean()

fig = px.bar(avg_marks, x="Subject", y="Marks",
             title="Average Marks by Subject",color="Subject")
fig.show()
```

```python
# 3D scatter plot
fig = px.scatter_3d(
    df,
    x="Study_Hours",
    y="Attendance",
    z="Marks",
    color="Subject",
    title="3D Student Performance"
)
fig.show()
```

```python
# 3D Line plot
fig = px.line_3d(
    df,
    x="Study_Hours",
    y="Attendance",
    z="Marks",
    color="Subject",
    title="3D Student Line Plot"
)

fig.show()
```

```python
# 3D Bubble Plot

fig = px.scatter_3d(
    df,
    x="Study_Hours",
    y="Attendance",
    z="Marks",
    size="Marks",
    color="Subject",
    hover_name="Student",
    title="3D Bubble Plot"
)

fig.show()
```

```python
 #surface Plot
import pandas as pd
import plotly.graph_objects as go
import numpy as np
x = np.array(df["Study_Hours"])
y = np.array(df["Attendance"])
z = np.array(df["Marks"])
fig = go.Figure(data=[
    go.Mesh3d(x=x, y=y, z=z, intensity=z)
])

fig.show()
```

```python
# Mesh Grid
x = np.linspace(0, 10, 20)
y = np.linspace(0, 10, 20)

X, Y = np.meshgrid(x, y)
Z = X + Y  # Example surface values

fig = go.Figure(data=[
    go.Surface(x=X, y=Y, z=Z)
])

fig.show()
```

```python
# Histogram 3D Plot
data = df["Study_Hours"].value_counts().sort_index()
fig = go.Figure(data=[
    go.Scatter3d(
        x=data.index,
        y=[0] * len(data),
        z=data.values,
        mode="markers+text",
        text=data.values,
        marker=dict(size=8),
        name="Student Count"
    )
])
fig.update_layout(
    title="3D Histogram-Style Plot: Students by Study Hours",
    scene=dict(
        xaxis_title="Study Hours",
        yaxis_title="Depth",
        zaxis_title="Number of Students"
    )
)

fig.show()
```

```python
data = df["Subject"].value_counts().reset_index()
data.columns = ["Subject", "Students"]
fig = px.pie(
    data,
    names="Subject",
    values="Students",
    title="Subject-wise Student Pie Chart",
    hole=0
)

fig.update_traces(
    pull=[0.05, 0.05, 0.05, 0.05],
    textinfo="label+percent"
)

fig.show()
```

```python
# Mesh Plot
from scipy.spatial import Delaunay

x = df["Study_Hours"].to_numpy()
y = df["Attendance"].to_numpy()
z = df["Marks"].to_numpy()

tri = Delaunay(np.column_stack((x, y)))

fig = go.Figure(go.Mesh3d(
    x=x, y=y, z=z,
    i=tri.simplices[:, 0],
    j=tri.simplices[:, 1],
    k=tri.simplices[:, 2],
    intensity=z,
    colorscale="Viridis"
))

fig.update_layout(title="3D Mesh Plot")
fig.show()
```
